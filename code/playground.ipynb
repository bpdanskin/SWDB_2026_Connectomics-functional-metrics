```python
import zarr, numpy as np
from pathlib import Path

def centroids_from_plane(zarr_path, plane_key):
    """Return list of (roi_id, cx_px, cy_px) from either mask format.
    Reads directly from zarr — avoids loading traces via pynwb."""
    z = zarr.open(str(zarr_path), mode='r')
    rt = z['processing'][plane_key]['image_segmentation']['roi_table']
    roi_ids = rt['roi'][:]
    
    records = []
    if 'pixel_mask' in rt:
        # Flat ragged array; pixel_mask_index[i] is the end of ROI i's slice
        pm_flat = rt['pixel_mask'][:]          # (total_pixels,) structured [('x','y','weight')]
        pm_idx  = rt['pixel_mask_index'][:]    # (n_rois,) cumulative end indices
        starts = np.concatenate([[0], pm_idx[:-1]])
        for i, (start, end) in enumerate(zip(starts, pm_idx)):
            px = pm_flat[start:end]
            cx = float(px['x'].mean())
            cy = float(px['y'].mean())
            records.append((int(roi_ids[i]), cx, cy))
    
    elif 'image_mask' in rt:
        # 3D array (n_rois, H, W), uint8 binary; axes: [roi, row=y, col=x]
        masks = rt['image_mask'][:]            # load all at once (512×512 per ROI)
        for i, mask in enumerate(masks):
            yx = np.argwhere(mask)             # (n_nonzero, 2) — row, col
            if len(yx) == 0:
                records.append((int(roi_ids[i]), np.nan, np.nan))
            else:
                cy = float(yx[:, 0].mean())   # row = y
                cx = float(yx[:, 1].mean())   # col = x
                records.append((int(roi_ids[i]), cx, cy))
    
    return records
```

```python
session_nwb_path = vn.find_sessions(Path('/data/409828_V1DD_Filtered'))[0]

records = []
for sess_path in vn.find_sessions(Path('/data/409828_V1DD_Filtered')):
    with NWBZarrIO(sess_path, mode = 'r') as _io:
        nwb = _io.read()
        for plane_key in vn.list_planes(nwb):
            records.append(centroids_from_plane(sess_path, plane_key))
```
